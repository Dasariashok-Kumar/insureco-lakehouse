```python
%sql
-- Free Edition gives you catalog creation rights on your own metastore
CREATE CATALOG IF NOT EXISTS insurance_dev;
CREATE SCHEMA IF NOT EXISTS insurance_dev.bronze;
CREATE SCHEMA IF NOT EXISTS insurance_dev.silver;
CREATE SCHEMA IF NOT EXISTS insurance_dev.gold;

-- Volumes = Unity Catalog's way of storing raw FILES (replaces old DBFS /FileStore)
CREATE SCHEMA IF NOT EXISTS insurance_dev.raw_files;
CREATE VOLUME IF NOT EXISTS insurance_dev.raw_files.landing;
```

```python
# Read directly from the Volumne path insteaed of /FileStore
df_customers = spark.read.csv("/Volumes/insurance_dev/raw_files/landing/customers.csv",
                              header = True,
                              inferSchema= True                              
                              ) 
display(df_customers)

# Write as a MANAGED Delta table inside Unity Catalog
df_customers.write.format("delta").mode("overwrite").saveAsTable("insurance_dev.bronze.customers")

# Confirm
display(spark.sql("SELECT * FROM insurance_dev.bronze.customers")) 
display(spark.sql("DESCRIBE DETAIL insurance_dev.bronze.customers"))
display(spark.sql("DESCRIBE HISTORY insurance_dev.bronze.customers"))    
           
```
