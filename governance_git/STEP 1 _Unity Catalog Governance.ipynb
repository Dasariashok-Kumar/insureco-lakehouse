Apply GRANT Statements 

account users is a built-in group in Unity Catalog meaning "everyone in this account." In a company, you'd replace this with actual group names like `claims_team` or `actuary_team` created by an admin.

```python
%sql
-- Grant broad catalog usage (required before any schema-level grant works)
GRANT USE CATALOG ON CATALOG insurance_dev TO `account users`;


-- Grant read-only access to Gold layer (simulating "actuary_team" / business users)
GRANT USE SCHEMA ON SCHEMA insurance_dev.gold TO `account users`;
GRANT SELECT ON SCHEMA insurance_dev.gold TO `account users`;

-- Grant read access to Silver (simulating "claims_team")
GRANT USE SCHEMA ON SCHEMA insurance_dev.silver TO `account users`;
GRANT SELECT ON SCHEMA insurance_dev.silver TO `account users`;

-- Bronze stays MORE restricted — only modify-level for data engineers
GRANT USE SCHEMA ON SCHEMA insurance_dev.bronze TO `account users`;
GRANT SELECT, MODIFY ON SCHEMA insurance_dev.bronze TO `account users`;
```

Verify Grants Applied

```python
%sql

SHOW GRANTS ON SCHEMA insurance_dev.gold;
```

Test REVOKE

```python
%sql

-- practise revoking(undo) a grant
REVOKE MODIFY ON SCHEMA insurance_dev.bronze FROM `account users`;

-- Confirm it's a gone
SHOW GRANTS ON SCHEMA insurance_dev.bronze;
```
