```python
from pyspark.sql.functions import col

CATALOG = "insurance_dev"

df_bronze = spark.table(f"{CATALOG}.bronze.claims_job_run")
df_clean = df_bronze.filter((col("claim_amount") >0 ) & col("claim_id").isNotNull() & col("claim_type").isNotNull())

df_clean.write.format("delta").mode("overwrite").saveAsTable(f"{CATALOG}.silver.claims_job_run")

print(f" Task 2 complete: {df_clean.count()} clean rows written to silver.claims_job_run")
```
